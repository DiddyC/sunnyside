```python
from redfin_scraper import RedfinScraper
import pandas as pd
```

```python
df  = pd.DataFrame.from_dict({'zip_codes':['07306']})
df.to_csv(r'zip_code_path.csv')
```

```python
scraper = RedfinScraper()

```

```python
scraper.setup(zip_database_path='zip_code_path.csv' ,multiprocessing=False)
```

```python
scraper.scrape(city_states=['Jersey City'],
               zip_codes=['07306'], sold=False )#sale_period:str=None, lat_tuner:float=1.5, lon_tuner:float=1.5)
```
